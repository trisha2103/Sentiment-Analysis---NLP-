```python
!pip install transformers
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import requests
from bs4 import BeautifulSoup
import re
```

```python
tokenizer = AutoTokenizer.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')

model = AutoModelForSequenceClassification.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')
```

```python
token = tokenizer.encode('I hated this, absolutely the worst', return_tensors='pt')
```

```python
token
```

```python
result = model(token)
```

```python
torch.argmax(result.logits)
```

```python
r = requests.get('https://www.trustpilot.com/review/hotchocolatedesign.com')
soup = BeautifulSoup(r.text, 'html.parser')
regex = re.compile('.*typography_body.-*')
results = soup.find_all('p', {'class': regex})
reviews = [result.text for result in results]
```

```python
reviews[0]
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(np.array(reviews), columns=['reviews'])
```

```python
df.head()
```

```python
def sentiment_score(review):
    tokens = tokenizer.encode(review, return_tensors='pt')
    result = model(tokens)
    return int(torch.argmax(result.logits))+1
    
```

```python
sentiment_score(df['reviews'].iloc[1])
```
